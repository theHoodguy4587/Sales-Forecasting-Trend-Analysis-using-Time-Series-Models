**Load Data**

```python
import pandas as pd

df=pd.read_csv("..\data\Sample - Superstore.csv",encoding='latin1')
df.columns=df.columns.str.strip()
df['Order Date']=pd.to_datetime(df['Order Date'])


monthly_sales=(df.set_index('Order Date').resample('M')['Sales'].sum().to_frame())
monthly_sales.head()
```

**Creating Time Based Data**

```python
monthly_sales.index=pd.to_datetime(monthly_sales.index)
monthly_sales['month']=monthly_sales.index.month
monthly_sales['year']=monthly_sales.index.year

monthly_sales.head()
```

**Creating Lag Features**

```python
monthly_sales['lag_1']=monthly_sales['Sales'].shift(1)
monthly_sales['lag_2']=monthly_sales['Sales'].shift(3)
monthly_sales['lag_3']=monthly_sales['Sales'].shift(6)
```

**Rolling Statistics** 

```python
monthly_sales['rolling_mean_3']=monthly_sales['Sales'].shift(1).rolling(window=3).mean()
monthly_sales['rolling_mean_6']=monthly_sales['Sales'].shift(1).rolling(window=6).mean()
monthly_sales.head()
```

**Drop Missing Values**

```python
monthly_sales=monthly_sales.dropna()

monthly_sales.head()
```

```python
monthly_sales.reset_index()
monthly_sales.head()
```

**Saving the feature Engineered Data**

```python
monthly_sales.to_csv("..\data\monthly_sales_features.csv")
```
